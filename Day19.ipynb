```python
# Series = One-dimensional data structure

marks = pd.Series([78,82,90,95])

print("Student Marks Series")
print(marks)

# Real-life Example
temperature = pd.Series([35,37,34,38])

print("\nTemperature Data")
print(temperature)
```

```python
# Real-life Example: Product Inventory

data = [
["Laptop",70000],
["Phone",30000],
["Tablet",25000]
]

products = pd.DataFrame(
data,
columns=["Product","Price"]
)

print(products)
```
